```python
df.iloc[10,:]
```

```python
# I am keeping the error above in the notebook because I am sure that I will have to deal with this issue again in the future. I want to 
# document it and understand it better.
# The command `df['horsepower'].fillna(df['horsepower'].mode()[0], inplace=True)` failed because Copy-on-Write (CoW) is now enabled 
# by default in pandas. 
# Under these rules, modifying a single column extracted from a DataFrame using an inplace=True method is treated as a chained assignment 
# and is explicitly blocked to prevent accidental data corruption.
```

```python
df['horsepower'] = df['horsepower'].fillna(df['horsepower'].mode()[0])
```

```python
df.iloc[10,:]
```

```python
df[df['horsepower'].isna()].head()
```

## Q7. Sum of weights

1. Select all the cars from Asia
2. Select only columns `vehicle_weight` and `model_year`
3. Select the first 7 values
3. Get the underlying NumPy array. Let's call it `X`.
4. Compute matrix-matrix multiplication between the transpose of `X` and `X`. To get the transpose, use `X.T`. Let's call the result `XTX`.
6. Invert `XTX`.
7. Create an array `y` with values `[1100, 1300, 800, 900, 1000, 1100, 1200]`.
9. Multiply the inverse of `XTX` with the transpose of `X`, and then multiply the result by `y`. Call the result `w`.
9. What's the sum of all the elements of the result?

```python
# 1. All the cars from Asia
all_the_cars_from_asia = df[df['origin'] == 'Asia']
all_the_cars_from_asia.head()
```

```python
# 2. Select only columns vehicle_weight and model_year
selected_columns = all_the_cars_from_asia[['vehicle_weight', 'model_year']]
selected_columns.head()
```

```python
# 3. Select the first 7 values
first_seven_values = selected_columns.head(7)
first_seven_values
```

```python
# 4. Get the underlying NumPy array. Let's call it X.
# IntelliSense (the autocomplete of VS Code) suggested to use `X = first_seven_values.values`
# But when I went to the official documentation of Pandas https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.values.html, 
# it shows a warning that recommends to use `to_numpy()` instead of `values`. So I will use that instead.
X = first_seven_values.to_numpy()
X
```

```python
# 5. Compute matrix-matrix multiplication between the transpose of X and X. To get the transpose, use X.T. Let's call the result XTX.
# We're dealing with a Numpy object, so we have to look in the Numpy documentation.
# To transpose a matrix, we can use the `transpose()` method or the `.T` attribute. See https://numpy.org/devdocs/reference/generated/numpy.ndarray.T.html
# And for the matrix-matrix multiplication, we can use the `matmul()` method or the `@` operator.
# See https://numpy.org/doc/stable/reference/generated/numpy.matmul.html.
# It's interesting that it uses the `@` operator because Pandas also does as per https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.transpose.html
# But why? It's because both rely on standard Python functionality introduced in PEP 465, which added the dedicated infix 
# operator @ to the Python language itself.
# Both libraries implement Python’s native __matmul__ special method. This creates a unified experience across the entire Python data science ecosystem.
# See https://peps.python.org/pep-0465/
XTX = X.T @ X
XTX
```

```python
# 6. Invert XTX.
# We're dealing with a Numpy object, so we have to look in the Numpy documentation for the method to invert a matrix.
# This https://numpy.org/doc/stable/reference/generated/numpy.invert.html#numpy.invert is NOT the correct documentation 
# because it is for inverting bits, not matrices. It calculates the bitwise "NOT" operation on the integers or booleans of the input array, element-wise.
# But https://numpy.org/devdocs/reference/generated/numpy.linalg.inv.html computes the multiplicative inverse of a square matrix. 
# It is the correct documentation for our case.
import numpy as np
XTX_inv = np.linalg.inv(XTX)
XTX_inv
```

```python
# What if we were still dealing with a Pandas DataFrame instead of a Numpy array? How would we invert it?
# There is no method in Pandas to invert a DataFrame.
# Do NOT use the `pow()` method to invert a DataFrame by using the power of -1. It is not the correct method to do that. 
# See https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.pow.html
# It would perform element-wise exponentiation, not matrix inversion.
```

```python
# 7. Create an array y with values [1100, 1300, 800, 900, 1000, 1100, 1200].
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])
y
```

```python
# 8. Multiply the inverse of XTX with the transpose of X, and then multiply the result by y. Call the result w.
w = XTX_inv @ X.T @ y
w
```

```python
# 9. What's the sum of all the elements of the result?
w.sum().round(3)
```

Answer: 0.369
